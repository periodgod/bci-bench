```python
from pathlib import Path
import sys

BENCH = Path(r'E:\\世界机器人大赛\\wyu-bci-bench')
if str(BENCH / 'src') not in sys.path:
    sys.path.insert(0, str(BENCH / 'src'))

from wyu_bci_bench.data import load_dataset
from wyu_bci_bench.evaluate import cross_validate_eegnet, cross_validate_filterbank_lda, dataset_quality_report
from wyu_bci_bench.viz import plot_band_energy_heatmap, plot_class_band_topomap, plot_label_distribution, plot_psd
```

```python
recordings = Path(r'E:\\世界机器人大赛\\projects\\arm_bci_robot\\data\\recordings')
out = Path(r'E:\\世界机器人大赛\\projects\\arm_bci_robot\\data\\experiments\\wyu_bci_notebook')
out.mkdir(parents=True, exist_ok=True)

dataset = load_dataset(recordings, label_mode='all', channel_mode='motor11')
print('shape:', dataset.shape)
print('fs:', dataset.fs)
print('labels:', dataset.label_names)
print('channels:', dataset.channel_names)
print(dataset.summary())
```

```python
quality = dataset_quality_report(dataset.eeg, dataset.y, dataset.fs, dataset.channel_names, dataset.label_names)
quality
```

```python
plot_label_distribution(dataset.labels, out / 'label_distribution.png')
plot_psd(dataset.eeg, dataset.fs, out / 'psd.png', dataset.y, dataset.label_names)
plot_band_energy_heatmap(dataset.eeg, dataset.fs, dataset.channel_names, out / 'band_energy_heatmap.png')
plot_class_band_topomap(dataset.eeg, dataset.y, dataset.fs, dataset.channel_names, out / 'mu_class_difference_topomap.png')
```

```python
fbcsp_lda = cross_validate_filterbank_lda(
    dataset.eeg,
    dataset.y,
    dataset.fs,
    dataset.channel_names,
    dataset.label_names,
    folds=5,
)
fbcsp_lda
```
